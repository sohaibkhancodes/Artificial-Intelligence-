```python
import json
from datetime import datetime

class UserProfile:
    def __init__(self, name, role, location, skills):
        self.name = name
        self.role = role
        self.location = location
        self.skills = skills
        self.created_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    def display_profile(self):
        print("=" * 45)
        print(f"       USER PROFILE: {self.name.upper()}")
        print("=" * 45)
        print(f"📍 Location : {self.location}")
        print(f"💼 Role     : {self.role}")
        print(f"🛠️ Skills   : {', '.join(self.skills)}")
        print(f"📅 Created  : {self.created_at}")
        print("=" * 45)

    def export_to_json(self, filename="user_data.json"):
        data = {
            "name": self.name,
            "role": self.role,
            "location": self.location,
            "skills": self.skills,
            "created_at": self.created_at
        }
        with open(filename, "w") as file:
            json.dump(data, file, indent=4)
        print(f"\n✅ Profile data saved to '{filename}' successfully!")


# --- Execution Example ---
if __name__ == "__main__":
    # Define profile details
    user = UserProfile(
        name="Sohaib",
        role="Student & Tech Enthusiast",
        location="Islamabad, Pakistan",
        skills=["Python", "SQL", "Cybersecurity", "Database Management"]
    )

    # Display formatted profile
    user.display_profile()

    # Save output to a JSON file in Google Colab
    user.export_to_json()
```
