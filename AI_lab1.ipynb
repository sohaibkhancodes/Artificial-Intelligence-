```python
# 1. Output and variables
name, age, gpa, passed = "Sohaib", 22, 3.5, True
print(f"Name: {name}, Age: {age}, GPA: {gpa}, Passed: {passed}")
print(type(name), type(age), type(gpa), type(passed))

# 2. Operators
a, b = 17, 5
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)
print(a > b, a == b, a > 10 and b < 10)

# 3. Strings
s = "Hello Python"
print(s.upper(), s.lower(), s[0:5], s[::-1], len(s))
print(s.replace("Python", "Colab"), s.split())

# 4. List
nums = [5, 2, 9, 1]
nums.append(7)
nums.sort()
print(nums, max(nums), min(nums), sum(nums))

# 5. Tuple and Set
t = (1, 2, 3)
st = {1, 2, 2, 3, 3}
print(t[0], st)

# 6. Dictionary
student = {"name": "Sohaib", "age": 22}
student["city"] = "Islamabad"
for k, v in student.items():
    print(k, ":", v)

# 7. If / else
marks = 78
if marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
else:
    print("Grade C")

# 8. Loops
for i in range(1, 6):
    print(i, end=" ")
print()
n = 3
while n > 0:
    print("Countdown:", n)
    n -= 1

# 9. Function
def square(x):
    return x * x

print(square(6))

# 10. List comprehension
print([x**2 for x in range(1, 6)])

# 11. Exception handling
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")

# 12. Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        return f"I am {self.name}, {self.age} years old."

print(Person("Sohaib", 22).intro())
```
